# **Data Preproccessing**

```python
#converting date to date-time
df_clean = df.copy()
df_clean["Date"] = pd.to_datetime(df_clean["Date"])
print(df_clean.dtypes)
```

```python
#Remove Xylene as high missing percentage
df_clean = df_clean.drop(columns=["Xylene"])

print("New dataset shape:", df_clean.shape)
print("Remaining columns:")
print(df_clean.columns.tolist())
```

```python
#check missing values again
missing_summary_clean = pd.DataFrame({
    "Missing Count": df_clean.isnull().sum(),
    "Missing Percentage": (
        df_clean.isnull().sum() / len(df_clean) * 100
    ).round(2)
})

missing_summary_clean.sort_values(
    "Missing Percentage",
    ascending=False
)
```

```python
#Handle missing AQI
df_clean = df_clean.dropna(subset=["AQI"])
print("Shape after removing missing AQI:", df_clean.shape)
```

```python
#One, global mean for all cities isnt viable
pollutant_columns = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3",
    "Benzene", "Toluene"
]
# For e.x: If PM2.5 is missing for a Delhi observation, fill it using the median PM2.5 value of Delhi, rather than the median PM2.5 of the entire country.
for col in pollutant_columns:
    df_clean[col] = df_clean.groupby("City")[col].transform(
        lambda x: x.fillna(x.median())
    )

print("Remaining missing values:")
print(df_clean.isnull().sum())
```

```python
#Chronological(Time wise) Sorting of data within each city
df_clean = df_clean.sort_values(
    ["City", "Date"]
).reset_index(drop=True)

print(df_clean[["City", "Date"]].head(10))
```

```python
print("Remaining missing values:")
print(df_clean.isnull().sum())
```

```python
#Overall median fallback
#If a city has no available values at all for a particular pollutant, we'll use the overall median of that pollutant.
#This prevents us from losing thousands of otherwise useful observations.
for col in pollutant_columns:
    overall_median = df_clean[col].median()

    df_clean[col] = df_clean[col].fillna(overall_median)
```

```python
print("Remaining missing values:")
print(df_clean.isnull().sum())
print("\nFinal cleaned shape:", df_clean.shape)
```

```python
#correlation amongst pollutants (collinearity)
correlation_matrix = df_clean[pollutant_columns + ["AQI"]].corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Pollutants and AQI")
plt.tight_layout()
plt.show()
```

The correlation analysis reveals several important relationships between AQI and the pollutant variables. CO shows the strongest positive linear correlation with AQI (0.68), followed by PM2.5 (0.66) and NO2 (0.53). This indicates that higher concentrations of these pollutants are generally associated with higher AQI values in the dataset. Other pollutants such as PM10, SO2, NOx and NO also show moderate positive relationships with AQI.

The analysis also identifies strong relationships among certain pollutants. NO and NOx have a high positive correlation of 0.79, while Benzene and Toluene show a correlation of 0.71. PM2.5 and PM10 also exhibit a moderate-to-strong positive correlation of 0.57. These relationships suggest that some pollutants may share related emission patterns or environmental conditions and also indicate potential multicollinearity that should be considered during model development.

NH3 shows relatively weak correlations with most other variables, including AQI, with values generally close to zero. This suggests that its linear relationship with the other measured pollutants and AQI is comparatively limited. Overall, the correlation analysis demonstrates that AQI is associated with multiple pollutants and that relationships between predictor variables should be considered when selecting features for machine learning.

Reason to perform correlation analysis?

=> We used correlation analysis to understand the linear relationships between pollutants and AQI, identify potentially important features, and detect highly correlated predictor variables that could introduce multicollinearity, especially for linear models.




```python
# multicollinearity analysis tells us whether predictor variables contain too much overlapping information.
# we'll be using VIF(Variance Inflation Factor) to quantify multicollinearity among the pollutant features.
#       VIF vs. Interpretation
#       1 =  no multicollinearity
#      1-5 = low to moderate
#      5-10 = High
#      >10 = very high


from statsmodels.stats.outliers_influence import variance_inflation_factor

# Select pollutant features
X_vif = df_clean[pollutant_columns].copy()

# Calculate VIF for each feature
vif_data = pd.DataFrame()
vif_data["Feature"] = X_vif.columns
vif_data["VIF"] = [
    variance_inflation_factor(X_vif.values, i)
    for i in range(X_vif.shape[1])
]

vif_data = vif_data.sort_values("VIF", ascending=False)

print(vif_data)
```

Variance Inflation Factor (VIF) analysis was performed to assess multicollinearity among the pollutant variables. The highest VIF was observed for NOx (3.45), followed by NO (2.96), Toluene (2.62), and Benzene (2.27). All remaining variables had VIF values below 2.20. Since all VIF values are below the commonly used threshold of 5, the predictor variables do not exhibit significant multicollinearity. Therefore, no pollutant feature was removed solely on the basis of VIF, allowing the complete set of pollutant variables to be considered during model development.
